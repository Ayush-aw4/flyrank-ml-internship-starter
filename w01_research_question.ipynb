## 2. The question: decision, action, cost of a wrong call

### Research Question

Can we use historical content performance and page characteristics to identify and rank website pages that deserve further investigation for potential content improvements?

### Decision

The decision is to determine which website content pages an SEO or content team should prioritize for review.

### Unit of Analysis

An individual website content page.

Each row in the starter dataset represents one pseudonymized content item.

### Expected Output

A ranked content-review queue containing:

- A priority score for each page.
- A priority category, such as high, medium or low.
- Reason codes explaining why a page received its score.
- Relevant performance signals supporting the recommendation.

### Who Will Use the Output?

SEO specialists and content managers will use the ranked queue to identify pages that deserve further investigation.

They can review the content and decide whether updating, restructuring or another improvement is appropriate.

### Action

The team will investigate high-priority pages first, examine their performance and content characteristics, and make an informed decision about possible improvements.

### Cost of a Wrong Recommendation

A false positive could cause the team to waste editorial time and resources updating a page that does not need improvement.

A false negative could cause an important declining page or potential improvement opportunity to be overlooked.

Both errors can lead to inefficient allocation of time and resources.

### Why Data and Machine Learning May Help

A website may contain thousands of content pages with multiple performance-related characteristics. Evaluating every page manually can be time-consuming.

Data analysis can help identify relationships between content characteristics and historical performance. Machine learning may help combine multiple signals and identify patterns that are difficult to express using simple rules.

However, the model must be compared against a transparent baseline. If a simple rule provides equally useful recommendations, a more complicated model may not be necessary.

The final system will support human decision-making rather than automatically deciding which pages must be updated.

```python
import pandas as pd

data_path = repo_root / "data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(data_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("Number of pages:", len(df))
print("Number of clients:", df["client_id"].nunique())

df.head()
```

```python
search_volume_corr = df["search_volume"].corr(
    df["impressions_90d"]
)

print(
    "Search volume vs impressions correlation:",
    round(search_volume_corr, 4),
)
```

```python
visible_pages = df[
    (df["impressions_90d"]>=100) &
    (df["avg_position"]>0)
].copy()

ctr_by_position = (
    visible_pages
    .groupby("position_tier")
    .agg(
        page_count = ("content_id", "count"),
        average_ctr = ("ctr", "mean")
    )
    .sort_values("average_ctr", ascending=False)
)
display(ctr_by_position.round(4))
```

```python
df["update_age_category"] = pd.cut(
    df["days_since_last_update"],
    bins=[-1, 30, 90, 180, float("inf")],
    labels=["Recently Updated", "Moderately Old", "Old", "Very Old"]
)

freshness_summary = (
    df.groupby("update_age_category", observed=False)
    .agg(
        page_count=("content_id", "count"),
        median_impressions=("impressions_90d", "median"),
        average_ctr=("ctr", "mean")
    )
)

display(freshness_summary.round(3))
```
